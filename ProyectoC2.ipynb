```python
%%writefile Aproximacion_numerica/entrada.py
def funcion1():
  while True:
    try:
      g = int(input("Inserte el grado de su polinomio: "))
      if g < 0:
        print("Error. El grado debe ser mayor o igual a 0")
        continue

      term = []
      datos = []
      for i in range(g+1):
        c = float(input(f"Inserte el coeficiente de x^{i}:"))
        if c == 0:
          continue
        datos.append((c,i))
        if i == 0:
          d = f"{c:g}"
        elif i == 1:
          if c == 1:
            d = f"x"
          elif c == -1:
            d = f"-x"
          else:
            d = f"{c:g}x"
        else:
          if c == 1:
            d = f"x^{i}"
          elif c == -1:
            d = f"-x^{i}"
          else:
            d = f"{c:g}x^{i}"
        term.append(d)
      return datos,term
    except ValueError:
      print("Error. Debes poner numeros")

def funcion2():
  while True:
    try:
      g = int(input("Inserte el grado de su polinomio: "))
      if g < 0:
        print("Error. El grado debe ser mayor o igual a 0")
        continue
      a = input(f"Inserte los coeficientes separados por , (recuerde que deben ser {g+1}):").split(",")
      if len(a) != g+1:
        print(f"Error. Debes poner exactamente {g+1} numeros")
        continue

      datos = []
      term = []
      for i in range(len(a)):
        c = (float(a[i]),i)
        datos.append(c)
      for j in datos:
        if j[0] == 0:
          continue
        elif j[0] == 1:
          if j[1] == 1:
            d = f"x"
          elif j[1] == 0:
            d = f"1"
          else:
            d = f"x^{j[1]}"
        elif j[0] == -1:
          if j[1] == 1:
            d = f"-x"
          elif j[1] == 0:
            d = f"-1"
          else:
            d = f"-x^{j[1]}"
        elif j[1] == 0:
          d = f"{j[0]:g}"
        elif j[1] == 1:
          d = f"{j[0]:g}x"
        else:
          d = f"{j[0]:g}x^{j[1]}"
        term.append(d)
      return datos,term
    except ValueError:
      print("Error. Debes poner numeros")

def mostrar_funcion(term):
  if not term:
    return "f(x) = 0"
  p = " + ".join(term).replace("+ -","- ")
  return f"f(x) = {p}"

def valores(datos):
  while True:
    try:
      v = list(map(float,input("Inserte el intervalo separado por ,: ").split(",")))
      if len(v) != 2:
        print("Error. Debes poner exactamente 2 numeros")
        continue
      if v[0] >= v[1]:
        print("Error. El primer numero debe ser menor al segundo")
        continue

      n = int(input("Inserte el numero de rectangulos (recuerde que debe ser par si se quiere aplicar Simpson): "))
      if n <= 0:
        print("Error. El numero de rectangulos debe ser mayor a 0")
        continue

      Δx = (v[1]-v[0])/n

      met = []
      tabla = []
      for i in range(n+1):
        suma = 0.0
        m = v[0] + Δx*i
        for j in datos:
          if j[0] == 0:
            continue
          elif j[1] == 0:
            suma += j[0]
          elif j[1] == 1:
            suma += j[0]*m
          else:
            suma += j[0]*(m**j[1])
        m_c = round(m, 6)
        suma_c = round(suma, 6)
        met.append((m_c,suma_c))
        xi = f"x{i} = {m_c}"
        fx = f"f(x{i}) = {suma_c}"
        tabla.append((xi,fx))
      return met,Δx,tabla,v,n
    except ValueError:
      print("Error. Debes poner numeros validos")
    except ZeroDivisionError:
      print("Error. El numero de rectangulos debe ser mayor a 0")
'''
def mostrar_valores(tabla):
  print("Valores:")
  for fila in tabla:
    print(fila)
'''
```

```python
%%writefile Aproximacion_numerica/metodos.py
def exceso(met,Δx):
  suma = 0
  for k in range(len(met)-1):
    suma += met[k][1]
  Area = Δx*suma
  print(f"El area por EXCESO es: {Area}")

def defecto(met,Δx):
  suma = 0
  for k in range(len(met)-1):
    suma += met[k+1][1]
  Area = Δx*suma
  print(f"El area por DEFECTO es: {Area}")

def p_medio(met,Δx,datos):
  p_med = []
  suma_p = 0.0
  for i in range(len(met)):
    suma = 0
    if i+1 >= len(met):
      break
    else:
      mi = (met[i][0]+met[i+1][0])/2
    for j in datos:
      if j[0] == 0:
        continue
      elif j[1] == 0:
        suma += j[0]
      elif j[1] == 1:
        suma += j[0]*mi
      else:
        suma += j[0]*(mi**j[1])
    p_med.append((mi,suma))
  for k in range(len(p_med)):
    suma_p += p_med[k][1]
  Area = Δx*suma_p
  print(f"\nEl area por el metodo del PUNTO MEDIO es: {Area}")

def tra(met,Δx):
  if len(met) == 1:
    print("Error. El numero de rectangulos debe ser mayor a 1")
    return
  suma = 0
  for k in range(len(met)):
    if k == 0:
      suma += met[k][1]
    elif k == len(met)-1:
      suma += met[k][1]
    else:
      suma += 2*met[k][1]
  Area = Δx*suma/2
  print(f"\nEl area por el metodo TRAPEZOIDAL es: {Area}")

def simp(met,Δx,n):
  if n%2 != 0:
    print("Error. El numero de rectangulos debe ser par")
    return
  suma = 0
  for i in range(len(met)):
    if i == 0:
      suma += met[0][1]
    elif i == len(met)-1:
      suma += met[len(met)-1][1]
    elif i%2 == 0:
      suma += 2*met[i][1]
    else:
      suma += 4*met[i][1]
  Area = Δx*suma/3
  print(f"\nEl area por el metodo de SIMPSON es: {Area}")
```

```python
%%writefile Aproximacion_numerica/errores.py
def der(datos):
  der = []
  for i in datos:
    if i[1] == 0:
      continue
    else:
      x = i[0]*i[1]
      fx = i[1]-1
    der.append((x,fx))

  if len(der) == 0:
    return [(0,0)]
  else:
    return der

def der_n(datos,n):
  resultado = datos
  for j in range(n):
    resultado = der(resultado)
  return resultado, n

def evaluar_polinomio(datos, x):
    suma = 0.0
    for coef, exp in datos:
        if coef != 0:
            suma += coef * (x ** exp)
    return suma

def integrar_polinomio_lista(polinomio):
    resultado = []

    for coef, exp in polinomio:
        nuevo_exp = exp + 1
        nuevo_coef = coef / nuevo_exp

        if nuevo_coef.is_integer():
            nuevo_coef = int(nuevo_coef)

        resultado.append((nuevo_coef, nuevo_exp))

    return resultado

def integral_definida(funcion, a, b):
    y = integrar_polinomio_lista(funcion)
    x = evaluar_polinomio(y, b) - evaluar_polinomio(y, a)
    print(f"El area bajo la curva es {x}")

def M(datos, v, c,x):
    a, b = v[0], v[1]
    paso = (b - a) / c
    max_val = 0.0

    if x == "2":
      derivada = der_n(datos, 2)[0]
    elif x == "4":
      derivada = der_n(datos, 4)[0]
    else:
      print("Error, mal escritura")

    for i in range(c + 1):
        x = a + i * paso
        val_abs = abs(evaluar_polinomio(derivada, x))

        if val_abs > max_val:
            max_val = val_abs

    print(f"El valor de M es: {max_val}")
    return max_val

def cerror_pm(m,Δx,n):
  En = float(m*((Δx*n)**3))/(24*(n**2))
  print(f"Cota de error Epm <= {En}")

def cerror_t(m,Δx,n):
  En = float(m*((Δx*n)**3))/(12*(n**2))
  print(f"Cota de error Et <= {En}")

def cerror_s(m,Δx,n):
  En = float(m*((Δx*n)**5))/(180*(n**4))
  print(f"Cota de error Es <= {En}")
```
